```python
import pandas as pd

df = pd.read_csv('../../data/raw/HomeC.csv')
print('Raw shape:', df.shape)
df.head()
```

```python
df_home = df.iloc[:, :6].dropna().copy()
df_home['time'] = df_home['time'].astype('int64')

print('Cleaned shape (first 6 columns):', df_home.shape)
print('Columns:', list(df_home.columns))
print('Missing values:\n', df_home.isna().sum())
df_home.head()
```

```python
import pandas as pd

# HomeC.csv records 1-minute intervals starting at 2016-01-01 00:00:00 local (US/Eastern)
start_dt = (
    pd.to_datetime(df_home['time'].iloc[0], unit='s', utc=True)
    .tz_convert('US/Eastern')
    .tz_localize(None)
)
df_home['datetime'] = pd.date_range(start=start_dt, periods=len(df_home), freq='1min')

# Drop raw integer time column and resample 1-minute readings to hourly averages (kW ≈ kWh per hour)
df_hourly = (
    df_home.drop(columns=['time'])
    .set_index('datetime')
    .resample('1h')
    .mean()
    .reset_index()
)

print('Hourly shape:', df_hourly.shape)
print('Date range:', df_hourly['datetime'].min(), '->', df_hourly['datetime'].max())
print('Missing values:\n', df_hourly.isna().sum())

clean_path = '../../data/processed/home_power_clean.csv'
df_hourly.to_csv(clean_path, index=False)
print('Saved updated dataset to:', clean_path)
df_hourly.head()
```

```python
import numpy as np
import pandas as pd

clean_path = '../../data/processed/home_power_clean.csv'
df_clean = pd.read_csv(clean_path, parse_dates=['datetime'])

# 1. Drop duplicate column (identical to 'use [kW]')
df_clean = df_clean.drop(columns=['House overall [kW]'], errors='ignore')

# 2. Engineer cyclical and calendar time features
hour = df_clean['datetime'].dt.hour
dow = df_clean['datetime'].dt.dayofweek
month = df_clean['datetime'].dt.month

df_clean['hour_sin'] = np.sin(2 * np.pi * hour / 24)
df_clean['hour_cos'] = np.cos(2 * np.pi * hour / 24)
df_clean['dow_sin'] = np.sin(2 * np.pi * dow / 7)
df_clean['dow_cos'] = np.cos(2 * np.pi * dow / 7)
df_clean['month_sin'] = np.sin(2 * np.pi * month / 12)
df_clean['month_cos'] = np.cos(2 * np.pi * month / 12)
df_clean['is_weekend'] = (dow >= 5).astype(int)

print('Updated shape:', df_clean.shape)
print('Columns:', list(df_clean.columns))
print('Missing values:\n', df_clean.isna().sum())

df_clean.to_csv(clean_path, index=False)
print('Updated single CSV:', clean_path)
df_clean.head()
```
